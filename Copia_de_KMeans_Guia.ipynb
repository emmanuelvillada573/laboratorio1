# Seminario de Ciencia de Datos
## Segmentación de Clientes con K-Means

Notebook estructurado con teoría, práctica, preguntas de evaluación y laboratorio guiado.

## Objetivos
- Comprender el aprendizaje no supervisado
- Aplicar K-Means
- Interpretar clusters y centroides
- Analizar resultados desde negocio

```python
# ===============================
# 1. IMPORTAR LIBRERÍAS
# ===============================

#COMPLETAR POR EL ESTUDIANTE
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

## Carga de Datos
Cargar el dataset de clientes.

```python
# COMPLETAR
df = pd.read_csv('/content/Mall_Customers - Mall_Customers.csv')
df.head()
```

## Selección de Variables
Usar: Annual Income y Spending Score.

```python
# COMPLETAR
X = df[["Annual Income (k$)", "Spending Score (1-100)"]]
```

## Método del Codo

```python
# COMPLETAR
inercia = []
for k in range(1, 11):
    modelo = KMeans(n_clusters=k, random_state=42)
    modelo.fit(X)
    inercia.append(modelo.inertia_)
```

## Entrenamiento del Modelo

```python
# COMPLETAR
kmeans = KMeans(n_clusters=4, random_state=42)
kmeans.fit(X)
df['Cluster'] = kmeans.labels_
```

## Visualización

```python
# COMPLETAR VISUALIZACIÓN
plt.scatter(df['Annual Income (k$)'], df['Spending Score (1-100)'], c=df['Cluster'])
plt.scatter(kmeans.cluster_centers_[:,0], kmeans.cluster_centers_[:,1])
```

## Respuestas

1.  **¿Por qué es aprendizaje no supervisado?**
    Es aprendizaje no supervisado porque el algoritmo K-Means agrupa los datos en clusters sin tener etiquetas predefinidas o una variable objetivo a predecir.

2.  **¿Qué es un centroide?**
    Un centroide es el punto central de un cluster. En K-Means, el centroide es la media de todos los puntos de datos asignados a ese cluster. Actúa como el representante o prototipo del cluster.

3.  **¿Qué sucede si K es muy alto?**
    Si K es muy alto, se crearán muchos clusters pequeños. Esto puede llevar a un sobreajuste  donde cada cluster contiene muy pocos puntos o incluso un solo punto, lo que dificulta la interpretación y la generalización de los resultados.

4.  **¿Para qué sirve el método del codo?**
    El método del codo  se utiliza para determinar el número óptimo de clusters  en K-Means. Consiste en graficar la inercia  en función del número de clusters . El codo en la gráfica, donde la disminución de la inercia se ralentiza significativamente, sugiere el valor óptimo de K.

5.  **Interprete los clusters desde negocio.**
    Para interpretar los clusters desde un punto de vista de negocio, se debe analizar las características promedio  de los clientes en cada cluster.
    *   **Cluster 1 morado :** Podrían ser clientes ahorradores o con ingresos limitados y poco gasto en el mall. Estrategia: Ofertas de valor, programas de fidelización con descuentos.
    *   **Cluster 2 verde :** Compradores impulsivos o con deudas pocos ingresos y muchas compras. Estrategia: Productos de temporada, microcréditos, programas de ahorro.
    *   **Cluster 3 amarillo :** Clientes pudientes pero cautelosos. Estrategia: Productos premium, experiencias exclusivas, asesoramiento personalizado.
    *   **Cluster 4 :** Clientes VIP, los más valiosos ganan mucho e invierten mucho. Estrategia: Trato personalizado, preventas exclusivas, gestión de cuenta.
   

## Conclusión
K-Means permite segmentar clientes y apoyar decisiones estratégicas basadas en datos.
